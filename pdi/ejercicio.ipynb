ejercicio N 3

```python
n = int(input("ingrese la cantidad de numero bleeees uwu que sea mas de 3 OwO"))
if n >= 3:
    nums = [int(input("numero: ")) for _ in range(n)]
    print(f"Mayor: {max(nums)}, Menor: {min(nums)}")
else:
    print("dije mas de 3 >:v")
```

ejercicio N 2

```python
n = int(input("introduce otro numero plis"))
if n > 1:
    for i in range(2, n):
        if n % i == 0:
            print(f"el numero {n} no es primo.")
            break
    else:
        print(f"el numero {n} es primo.")
else:
    print(f"el numero {n} no es primo.")
```
