BGR a escala de grisses (LISTO)

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Para la primera imagen
img_bgr = cv2.imread("img3.jpeg")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

fig, axs = plt.subplots(3, 4, figsize=(12, 10))

axs[0, 0].imshow(img_rgb)
axs[0, 0].set_title('Imagen original')

axs[0, 1].imshow(img_rgb[..., 0], cmap='gray')
axs[0, 1].set_title('Canal rojo')

axs[0, 2].imshow(img_rgb[..., 1], cmap='gray')
axs[0, 2].set_title('Canal verde')

axs[0, 3].imshow(img_rgb[..., 2], cmap='gray')
axs[0, 3].set_title('Canal azul')

# Para la segunda imagen
img_bgr1 = cv2.imread("img2.jpeg")
img_rgb1 = cv2.cvtColor(img_bgr1, cv2.COLOR_BGR2RGB)

axs[1, 0].imshow(img_rgb1)
axs[1, 0].set_title('Imagen original')

axs[1, 1].imshow(img_rgb1[..., 0], cmap='gray')
axs[1, 1].set_title('Canal rojo')

axs[1, 2].imshow(img_rgb1[..., 1], cmap='gray')
axs[1, 2].set_title('Canal verde')

axs[1, 3].imshow(img_rgb1[..., 2], cmap='gray')
axs[1, 3].set_title('Canal azul')

# Pa la tercera imagen
img_bgr2 = cv2.imread("img1.jpeg")
img_rgb2 = cv2.cvtColor(img_bgr2, cv2.COLOR_BGR2RGB)

axs[2, 0].imshow(img_rgb2)
axs[2, 0].set_title('Imagen original')

axs[2, 1].imshow(img_rgb2[..., 0], cmap='gray')
axs[2, 1].set_title('Canal rojo')

axs[2, 2].imshow(img_rgb2[..., 1], cmap='gray')
axs[2, 2].set_title('Canal verde')

axs[2, 3].imshow(img_rgb2[..., 2], cmap='gray')
axs[2, 3].set_title('Canal azul')


plt.tight_layout()
plt.show()
```

FALSO COLOR (LISTO)

```python
import cv2
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = [15, 5]

img = cv2.imread("img3.jpeg", cv2.IMREAD_COLOR)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
b, g, r = cv2.split(img)

plt.subplot(141)
plt.title("Original")
plt.imshow(img_rgb)
plt.axis("off")

plt.subplot(142)
plt.title("Rojo")
plt.imshow(r, cmap='Reds')
plt.axis("off")

plt.subplot(143)
plt.title("Verde")
plt.imshow(g, cmap='Greens')
plt.axis("off")

plt.subplot(144)
plt.title("Azul")
plt.imshow(b, cmap='Blues')
plt.axis("off")

plt.show()
```

formato de colores (LISTO)

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["figure.figsize"] = [15, 5]

img = cv2.imread("img1.jpeg", cv2.IMREAD_COLOR)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r = img[:, :, 2]
g = img[:, :, 1]
b = img[:, :, 0]

r_img = np.zeros_like(img_rgb)
r_img[:, :, 0] = r

g_img = np.zeros_like(img_rgb)
g_img[:, :, 1] = g

b_img = np.zeros_like(img_rgb)
b_img[:, :, 2] = b

plt.subplot(141)
plt.title("Original")
plt.imshow(img_rgb)
plt.axis("off")

plt.subplot(142)
plt.title("Rojo")
plt.imshow(r_img)
plt.axis("off")

plt.subplot(143)
plt.title("Verde")
plt.imshow(g_img)
plt.axis("off")

plt.subplot(144)
plt.title("Azul")
plt.imshow(b_img)
plt.axis("off")

plt.show()
```
