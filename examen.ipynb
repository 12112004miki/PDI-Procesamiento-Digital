```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('cawallo.jpeg')
imagen = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
altura, ancho, _ = imagen.shape
degradado = np.linspace(0, 1, altura, dtype=np.float32).reshape(-1, 1)
degradado = np.repeat(degradado, ancho, axis=1)

imagen = imagen.astype(np.float32) / 255.0
img_deg = imagen * degradado[..., np.newaxis]
img_deg = (img_deg * 255).astype(np.uint8)

plt.imshow(img_deg)
plt.show()
```

EJERCICIO N4

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('cawallo.jpeg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

alto, ancho = img.shape[:2]

mascara = np.zeros((alto, ancho))
puntos = np.array([[(ancho // 2, alto // 2 - 60),
                    (ancho // 2 - 60, alto // 2 + 60),
                    (ancho // 2 + 60, alto // 2 + 60)]])
cv2.fillPoly(mascara, puntos, 255)
img_mascara = cv2.bitwise_and(img, img, mask=mascara)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.subplot(1, 2, 2)
plt.imshow(img_mascara)
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LinearSegmentedColormap

img_bgr = cv2.imread("cawallo.jpeg")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

ventana = 15
factor = 2
imagen_umbr = cv2.adaptiveThreshold(img_gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, ventana, factor)

def solarizar(img, umbral):
    return np.where(img < umbral, img, 255 - img)

img_solarizada = solarizar(img_bgr, 100)
imagen_negativo = 255 - img_solarizada

plt.figure(figsize=(10, 5))

plt.subplot(2, 3, 1)
plt.title("Original")
plt.imshow(img_rgb)

plt.subplot(2, 3, 2)
plt.title("Solarizada")
plt.imshow(cv2.cvtColor(img_solarizada, cv2.COLOR_BGR2RGB))

plt.subplot(2, 3, 3)
plt.title("Posterizada")
plt.imshow(cv2.cvtColor(imagen_negativo, cv2.COLOR_BGR2RGB))

plt.subplot(2, 3, 4)
plt.title('Imagen con umbral adaptativo')
plt.imshow(imagen_umbr, cmap='gray')

plt.subplot(2, 3, 5)
plt.title("Punto de interés")
plt.imshow(img_rgb[100:205, 100:205])

plt.show()
```
