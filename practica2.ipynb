```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

pato = cv2.imread('pato.jpg', 0)
mask = cv2.imread('puntos.png', 0)
mask = cv2.resize(mask, (pato.shape[1], pato.shape[0]))
puntos_bin = cv2.threshold(mask, 127, 255, cv2.THRESH_BINARY_INV)[1]

img = cv2.bitwise_and(pato, pato, mask=puntos_bin)
img = np.where(puntos_bin == 0, 255, img).astype(np.uint8)

plt.figure(figsize=(10, 5))
plt.subplot(1,3,1)
plt.imshow(pato, cmap='gray')
plt.title('Pato')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(puntos, cmap='gray')
plt.title('mask')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img, cmap='gray')
plt.title('Pato con Puntos')
plt.axis('off')

plt.show()
```

El operador de Canny es un algoritmo de detección de bordes utilizado en procesamiento de imágenes este algoritmo identifica los bordes en una imagen mediante una serie de pasos:

reducción de ruido, cálculo de gradientes, supresión de no-máximos y, finalmente, aplicación de umbrales para obtener los bordes finales

En el contexto del reconocimiento facial, Canny se utiliza para destacar las características principales de las caras y sus bordes, lo que facilita el análisis.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('am.jpeg', cv2.IMREAD_COLOR)
img_gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(img_gray, (7, 7), 1.5)

bajos = 30
altos = 100

res = cv2.Canny(blur, bajos, altos)

plt.figure(figsize=(10,10))
plt.subplot(1, 2, 1)
plt.title("Imagen Original")
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
plt.subplot(1, 2, 2)
plt.title("Bordes Canny")
plt.imshow(res, cmap='gray')
plt.show()
```
