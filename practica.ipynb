PATRONES DE MOIRE

Los patrones de moiré son interferencias visuales que se producen cuando dos o más patrones repetitivos se superponen. Estos patrones crean un efecto ondulado o de franjas que no está presente en ninguno de los patrones individuales. El efecto se debe a la interacción de las ondas de los patrones, lo que resulta en un nuevo patrón de interferencia.


ALIASING

El aliasing es un fenómeno que ocurre en la digitalización de señales, especialmente en audio e imágenes, cuando la frecuencia de muestreo es demasiado baja, lo que causa la distorsión de la señal y la aparición de frecuencias falsas.

IMAGEN N1

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucionMediaHorizontal(imagen,tamkernel):
    kernel=np.ones((tamkernel,1),np.float32)/(tamkernel)
    imagen_convolucionada=cv2.filter2D(imagen,-1,kernel)
    return imagen_convolucionada


imagen=cv2.imread('prac2.jpeg')
img_rgb=cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB)

igmRest_r=convolucionMediaHorizontal(img_rgb[:,:,0],15)
igmRest_g=convolucionMediaHorizontal(img_rgb[:,:,1],15)
igmRest_b=convolucionMediaHorizontal(img_rgb[:,:,2],15)

imgColor=np.stack([igmRest_r,igmRest_g,igmRest_b],axis=2)

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(imgColor)
plt.axis('off')
plt.show()
```

Imagen N2

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convMediaHor(imagen, tamanioK):
    kernel = np.ones((1, tamanioK), np.float32) / tamanioK
    kernel2 = np.ones((tamanioK, 1), np.float32) / tamanioK
    imgFin = cv2.filter2D(imagen, -1, kernel)
    imgFin = cv2.filter2D(imgFin, -1, kernel2)
    return imgFin

imagen = cv2.imread("prac3.jpeg")

gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

img = convMediaHor(gris, 8)

plt.subplot(1,2,1)
plt.axis("off")
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))

plt.subplot(1,2,2)
plt.axis("off")
plt.imshow(img)
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convMediaHor(imagen, tamanioK):
    kernel = np.ones((1, tamanioK), np.float32) / tamanioK
    kernel2 = np.ones((tamanioK, 1), np.float32) / tamanioK
    imgFin = cv2.filter2D(imagen, -1, kernel)
    imgFin = cv2.filter2D(imgFin, -1, kernel2)
    return imgFin

imagen = cv2.imread("prac1.jpeg")

gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

img = convMediaHor(gris, 50)

plt.figure(figsize=(15, 10))
plt.subplot(121)
plt.title("Imagen Original")
plt.axis("off")
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))

plt.subplot(122)
plt.axis("off")
plt.imshow(img)
plt.show()
```
